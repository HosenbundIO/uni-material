```python
def quersumme(x):
    quersumme = 0
    for ziffer in str(x):
        quersumme += int(ziffer)
    return quersumme

for i in range(1000):
    if (i % 6 == 0) and (i % 7 == 0) and (i % 4 > 0) and (quersumme(i) == 6):
        print(i)
```

```python
def middle_square(x):
    return (x * x) // 0x100 % 0x10000
```

```python
def hase_igel():
    for i in range(0x10000):
        # Aktualisierung
        hase = middle_square(middle_square(i))
        igel = middle_square(i)

        while hase != igel:
            # Berechnung/Aktualisierung
            hase = middle_square(middle_square(hase))
            igel = middle_square(igel)

        if igel != middle_square(igel):
            print(i)
```
