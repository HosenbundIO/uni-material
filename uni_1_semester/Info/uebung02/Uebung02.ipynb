### Größter gemeinsamer Teiler (ggT)

**Das ggT-Problem.&ensp; Gegeben:** Zwei positive ganze Zahlen $a, b \in \mathbb{Z}$ mit $a, b > 0$. 

**Gesucht**: Eine positive ganze Zahl $d \in \mathbb{Z}$ mit $d > 0$ und folgenden Eigenschaften:

* $d$ teilt $a$ und $b$ ohne Rest,
* es gibt **keine** ganze Zahl $d' \in \mathbb{Z}$ mit $d' > d$, die $a$ und $b$ ohne Rest teilt.

---

### Der euklidische Algorithmus (klassisch)

**Eingabe:** Zwei ganze Zahlen $a, b$ mit $a \ge b \ge 0$.

**Initialisierung:**
$$
x \leftarrow a, \quad y \leftarrow b
$$

**Iteration:** Wiederhole Folgendes, solange $y \ne 0$:

* Falls $x > y$, dann $x \leftarrow x - y$.
* sonst $y \leftarrow y - x$.

**Abschluss:**
$$
d \leftarrow x
$$

**Rückgabe:** $d$

*Hinweis:* $A \leftarrow B$ bedeutet: speichere den aktuellen Wert von $B$ in $A$.

---

### Bemerkung.

Die Formatierung muss nicht exakt übereinstimmen. Es reicht den Inhalt korrekt wiederzugeben.

Den Pfeil nach links wird durch den Befehl 

`\leftarrow`

erzeugt.

```python
# Rumpf für Aufgabe 3.1
# Hinweis: Um eine Zahl in einen Text umzuwandeln gibt es die funktion "str(zahl)"; um einen Text in eine Zahl umzuwandlen gibt es die funktion "int(text)"
def quersumme(x):
    quersumme = 0
    for ziffer in str(x):
        quersumme += int(ziffer)
    print(quersumme)
    return quersumme

assert quersumme(42) == 6
assert quersumme(3678) == 24
assert quersumme(6947859) == 48
```

```python
def iterierte_quersumme(x):
    step = 1
    while len(str(x)) > 1:
        quersumme = 0
        for ziffer in str(x):
            quersumme += int(ziffer)
        x = quersumme
        print(f'{step}. {quersumme}')
        step += 1
    return x

assert iterierte_quersumme(24) == 6
assert iterierte_quersumme(8763) == 6
assert iterierte_quersumme(9587496) == 3
```
