```python
def draw_square_spiral(t, steps, start_length, increment):
    for i in range(steps):
        t.forward(start_length)
        t.left(90)
        start_length += increment
```

```python
def read_words(filename):
    with open(filename, encoding="utf-8") as f:
        for line in f:
            for word in line.split():
                yield word
```

```python
words = {}
sum = 0

for i in read_words("woerter2.txt"):
    words[i] = words.get(i, 0) + 1
    sum += len(i)


print(f"Dict wort --> anzahl_vorkommen: {words}")

sorted_dic = sorted(words.items(), key=lambda item: item[1], reverse=True)
print(f"Top 10: {sorted_dic[:10]}")
print(f"Last 10: {sorted_dic[-10:]}")

print(f"Durchschnittliche Länge aller gelesenen Wörter: {sum / len(words)} Buchstaben pro Wort")

length_dic = {}
for i in words.keys():
    length = len(i)
    if length not in length_dic:
        length_dic[length] = []

    length_dic[length].append(i)

print(f"Dict Wortlänge --> Wert: {length_dic}")

alpha_words = {word: count for word, count in words.items() if word.isalpha()}
print(f"Dict nur mit alphabetischen Zeichen: {alpha_words}")
```

```python
def christmas_tree(n, ornament="*"):

    lines = []
    if n < 3 or n % 2 == 0:
        raise ValueError("n muss eine ungerade Zahl >= 3 sein.")
    
    for i in range(1, n + 1, 2):
        spaces = (n - i) // 2

        if i == 1: 
            lines.append(" " * spaces + "+")

        lines.append(" " * spaces + i * ornament)

    trunk_space = (n - 1) // 2
    lines.append(" " * trunk_space + "|")

    gift_spaces = (n - 3) // 2
    if gift_spaces < 0: gift_spaces = 0
        
    lines.append(" " * gift_spaces + "🎁🎁🎁")


    return "\n".join(lines)

```

```python
display_tree(christmas_tree(29))
```
