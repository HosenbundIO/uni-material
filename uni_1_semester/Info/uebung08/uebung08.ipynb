```python
def levy_next(curve):
    return ['+'] + curve + ['-'] + ['-'] + curve + ['+']
```

```python
def levy_flr(curve):
    first_f = curve.index("F")
    last_f = len(curve) - 1 - curve[::-1].index('F')
    part = curve[first_f:last_f + 1]
    
    result = []
    result.append('F')
    counter = 0
    for char in part[1:]:
        if char == "+":
            counter += 1
        elif char == '-':
            counter -= 1
        elif char == 'F':
            
            steps = counter // 2
            if steps == 1:
                result.append('R')
            elif steps == -1:
                result.append('L')

            result.append('F')
            counter = 0
    return result

```

```python
def write_file(filename, levyc):
    """
    Schreibt die Zeichenkette in eine Datei.
    """
    with open(filename, "w", encoding="utf-8") as f:
        f.write(levyc)
```

```python
def levy_generator(k):
    curve = ['F']
    for i in range(k):
        curve = levy_next(curve)

    flr_curve = levy_flr(curve)

    curve_string = "".join(flr_curve)
    write_file("levy_curve.txt", curve_string)
```

```python
levy_generator(10)
```

```python
# ------------------------------------------------------------
# Zeichenkette aus Datei lesen
# ------------------------------------------------------------

def read_commands(filename):
    """
    Liest die Zeichenkette der Levy-C-Kurve aus einer Datei.
    Erlaubt sind nur die Zeichen 'F', 'L' und 'R'.
    """
    with open(filename, encoding="utf-8") as f:
        return "".join(c for c in f.read() if c in "FLR")
```

```python
def jospehus_recursive(circle, current, round, step):
    
    if round > len(circle):
        return
    
    n = len(circle)
    count = 0
    pos = current

    while count < step:
        if circle[pos] == 0:
            count += 1
        if count < step:
            pos = (pos + 1) % n
    
    circle[pos] = round
    jospehus_recursive(circle, (pos + 1) % n, round + 1, step)
```

```python
def josephus_permutation(length, step):
    circle = [0] * length

    jospehus_recursive(circle, current=0, round=1, step=step)
    return circle
```

```python
def print_josephus(length, step):
    l = josephus_permutation(length, step)

    print(" ".join(str(i+1) for i in range(length)))
    print(" ".join(str(x) for x in l))
```

```python
print_josephus(6, 3)
```
